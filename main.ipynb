## 6. Write a function that returns the factorial of a number.

```python
def factorial(n):
    if n <0:
        return
    sum =1
    for number in range (1,n+1): 
        sum*= number
    return sum

print(factorial(6))
print(factorial(1))
print(factorial(0))
```

```python
#factorial formula
#n! = n * ( n - 1 ) *...*1

# This code defines a function called "factorial" which takes an input "n". The function uses a for loop to iterate through the range of numbers 
# from 1 to n+1. For each number in that range, it multiplies the current value of x by the number in the range. At the end of the loop, 
# the function returns the final value of x, which will be the factorial of the input number "n".

# The Factorial of a positive integer n is the product of all positive integers less than or equal to n. 
# For example, the factorial of 6 (written "6!") is 6 * 5 * 4 * 3 * 2 * 1 = 720.

# So this function takes an input of any positive integer, and returns the factorial of that number.
```

```python
# This will test your function 
test_factorial(factorial)
```

## 7. Write a function that takes a list and returns a list of the unique values.

`NOTE: You cannot use set. 🤔`

```python
def unique(lst_un):
    unique = []
    for val in lst_un:
        if val in unique:
            continue
        else:
            unique.append(val)
    return unique

print(unique(["Alice", "Charlie", "Dave", "Bob", "Charlie", "Dave"])) 
```

## 8. Write a function that returns the mode of a list, i.e.: the element that appears the most times.
`NOTE: You should not use count... 🧐`

```python
def mode_counter(arr):
    frequency = {}

    for item in arr:
        if item in frequency:
            frequency[item] += 1
        else:
            frequency[item] = 1

    mode = None
    highest_freq = 0

    for item, freq in frequency.items():
        if freq > highest_freq:
            highest_freq = freq
            mode = item

    return mode


# Examples
print(mode_counter([1, 2, 2, 3, 3, 3, 4]))      
print(mode_counter(["a", "b", "a", "c", "a"])) 
```

```python
# This will test your function 
test_mode(mode_counter)
```

## 9. Write a function that calculates the standard deviation of a list.
`NOTE: Do not use any libraries or already built functions. 😉`

!!!!!! How it's calculated, step by step:

1.Find the mean of the data.

2.For each value, find how far it is from the mean, and square that difference (squaring makes everything positive, so distances above and below the mean don't cancel out).

3.Average those squared differences — this is called the variance.

4.Take the square root of the variance to get back to the original units — that square root is the standard deviation

```python
def st_dev(list_sd):
    n = len(list_sd)

    # Step 1: calculate the mean
    total = 0
    for num in list_sd:
        total += num
    mean = total / n

    # Step 2: sum of squared differences from the mean
    squared_diff_sum = 0
    for num in list_sd:
        squared_diff_sum += (num - mean) ** 2

    # Step 3: variance — divide by (n - 1) for SAMPLE standard deviation,
    # matching statistics.stdev()
    variance = squared_diff_sum / (n - 1)

    # Step 4: square root of variance (no math.sqrt)
    result = variance ** 0.5

    return result

print(st_dev([2, 4, 4, 4, 5, 5, 7, 9]))  
print(st_dev([10, 12, 23, 23, 16, 23, 21, 16]))
```

```python
# This will test your function 
test_stdev(st_dev)
```

## 10. Write a function to check if a string is a pangram, i.e.: if it contains all the letters of the alphabet at least once. Mind that the strings may contain characters that are not letters.

```python
def pangram(string):
    alphabet = set("abcdefghijklmnopqrstuvwxyz")
    return alphabet <= set(string.lower())

print(pangram("The quick brown fox jumps over the lazy dog"))
print(pangram("Things are getting complecated"))
```

## 11. Write a function that receives a string of comma separated words and returns a string of comma separated words sorted alphabetically.

`NOTE: You may use sorted but not split and definitely no join! 🤪`

```python
def sort_alpha(string):
    words = []
    current_word = ""
    for char in string:
        if char == ",":
            words.append(current_word)
            current_word = ""
        else:
            current_word += char
    words.append(current_word)

    words = sorted(words)

    result = ""
    for i in range(len(words)):
        result += words[i]
        if i != len(words) -1:
            result += ","

    return result

print(sort_alpha("Salma,Luis,Zina,Jose"))     
```

```python
# This will test your function 
test_alpha(sort_alpha)
```

## 12. Write a function to check if a given password is strong (at least 8 characters, at least one lower case, at least one upper case, at least one number and at least one special character). It should output True if strong and False if not.

```python
def check_pass(password):
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False

    special_chars = "<>?/{}~|!@#$%^&*()-_+="

    if len(password) < 8:
        return False

    for char in password:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_digit = True
        elif char in special_chars:
            has_special = True

    return has_lower and has_upper and has_digit and has_special


# Examples
print(check_pass("Abcdef1!"))   
print(check_pass("abcdefg1"))   
print(check_pass("Short1!"))    
print(check_pass("ALLUPPER1!")) 
```

```python
# This will test your function 
test_pass(check_pass)
```
